```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D


def log_k_of_theta(theta: np.ndarray, n: int) -> np.ndarray:
    return n * np.log(1.0 / np.sin(theta))


def plot_log_k_vs_theta(save_path: Path, epsilon: float = 0.1, num_points: int = 2000) -> None:
    theta = np.linspace(1e-4, np.pi / 2, num_points)

    dims = [768, 4096]
    colors = ["tab:blue", "tab:orange"]

    plt.figure(figsize=(12, 8))

    for n, color in zip(dims, colors):
        log_k = log_k_of_theta(theta, n)
        plt.plot(theta, log_k, linewidth=3, color=color, label=fr"$n={n}$")

    plt.axvline(x=epsilon, color="black", linestyle="--", linewidth=2.5)

    plt.xlabel(r"angle $\phi$", fontsize=28)
    plt.ylabel(r"$\log \Lambda = n \log\!\left(\frac{1}{\sin(\phi)}\right)$", fontsize=28)

    xticks = [0, np.pi / 8, np.pi / 4, 3 * np.pi / 8, np.pi / 2]
    xtick_labels = [
        r"$0$",
        r"$\frac{\pi}{8}$",
        r"$\frac{\pi}{4}$",
        r"$\frac{3\pi}{8}$",
        r"$\frac{\pi}{2}$",
    ]
    plt.xticks(xticks, xtick_labels, fontsize=24)
    plt.yticks(fontsize=24)

    plt.grid(True, linestyle="--", alpha=0.35)

    handles, _ = plt.gca().get_legend_handles_labels()
    epsilon_handle = Line2D(
        [0], [0],
        color="black",
        linestyle="--",
        linewidth=2.5,
        label=fr"$\phi = {epsilon}$",
    )
    handles.append(epsilon_handle)

    plt.legend(handles=handles, fontsize=22, title_fontsize=24)

    plt.tight_layout()
    plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.show()
    plt.close()

    print(f"Saved plot to: {save_path}")


if __name__ == "__main__":
    plot_log_k_vs_theta(Path("log_k_vs_theta_residual_dims.png"))
```
